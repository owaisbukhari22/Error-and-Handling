### Q4
Ek file `numbers.txt` banayein aur us mein 1 se 5 tak numbers line-by-line write karein.

```python
with open("numbers.txt", "w") as f:
    for i in range(1, 6):      # 1 se 5 tak
        f.write(str(i) + "\n")  # har number ke baad nayi line
print("numbers.txt ban gayi.")
```

### Q5
`numbers.txt` file ko read karein aur tamam numbers ka sum (total) calculate karke print karein.

```python
total = 0
with open("numbers.txt", "r") as f:
    for line in f:
        total += int(line.strip())   # strip() se \n hat jata hai
print("Sum =", total)
```

### Q6
Ek Python program likhein jo kisi file mein total lines count karke bataye.

```python
filename = "numbers.txt"   # yahan apni koi bhi file ka naam likh sakte hain

with open(filename, "r") as f:
    lines = f.readlines()

print("Total lines:", len(lines))
```

### Q7 (Medium)
`with open()` statement ka istemal karte hue ek file `data.txt` read karein aur explain karein ki `with` ka kya faida hai.

```python
# pehle data.txt bana lete hain taake error na aaye
with open("data.txt", "w") as f:
    f.write("Yeh data.txt ka content hai.\nDusri line.")

# with statement ke sath read
with open("data.txt", "r") as f:
    print(f.read())

print("File closed hai?", f.closed)   # True
```

### Q8 (Medium)
Ek file `sample.txt` se read karke uski saari lines ko ek Python list mein convert karein (`readlines()` ka use karke).

```python
with open("sample.txt", "w") as f:       # pehle sample file bana lein
    f.write("Line 1\nLine 2\nLine 3\n")

with open("sample.txt", "r") as f:
    lines = f.readlines()

print(lines)   # ['Line 1\n', 'Line 2\n', 'Line 3\n']
```

### Q9
Ek file `source.txt` ka data read karke `destination.txt` file mein copy karein.

```python
with open("source.txt", "w") as f:        # source file bana lein
    f.write("Yeh source file ka data hai.\nCopy hona chahiye.")

with open("source.txt", "r") as src:
    data = src.read()

with open("destination.txt", "w") as dest:
    dest.write(data)

with open("destination.txt", "r") as f:
    print(f.read())
```

### Q10

### Q11

### Q12 (Easy)
User se number input lein. Agar user text enter kar de toh `ValueError` handle karke custom error message print karein.

```python
try:
    num = int(input("Ek number enter karein: "))
    print("Aap ne number enter kiya:", num)
except ValueError:
    print("Error: Yeh number nahi hai! Baraye meherbani sirf number likhein.")
```

### Q13 (Easy)
Aisi file open karne ki koshish karein jo exist nahi karti (`non_existent_file.txt`). `FileNotFoundError` ko `try-except` se handle karein.

```python
try:
    with open("non_existent_file.txt", "r") as f:
        print(f.read())
except FileNotFoundError:
    print("Error: File nahi mili! Naam ya path check karein.")
```

### Q14
Ek program banayein jisme `finally` block ho, jo humesha execute ho chahe error aaye ya na aaye.

```python
try:
    x = 10 / 0
except ZeroDivisionError:
    print("Error aaya!")
finally:
    print("Finally block humesha chalta hai.")

print("---- Ab bina error ke ----")

try:
    x = 10 / 2
    print("Result:", x)
except ZeroDivisionError:
    print("Error aaya!")
finally:
    print("Finally block humesha chalta hai.")
```

### Q15
Ek list `my_list = [10, 20, 30]` banayein. Invalid index (e.g., index 5) access karein aur `IndexError` catch karein.

### Q16

### Q17
Ek `try-except` block likhein jo Multiple Exceptions catch kar sake (`ZeroDivisionError` aur `ValueError`).

### Q18
`else` block ka istemal `try-except` ke sath karein jo sirf tab chale jab koi error na aaye.

```python
try:
    result = 10 / 2
except ZeroDivisionError:
    print("Error aaya!")
else:
    print("Koi error nahi aaya. Result =", result)   # sirf tab chalega
```

### Q19
`raise` keyword ka use karke age validation ka function banayein. Agar age 18 se kam ho toh custom `ValueError("Age must be 18 or older")` raise karein.

```python
def validate_age(age):
    if age < 18:
        raise ValueError("Age must be 18 or older")
    return "Age valid hai."

try:
    print(validate_age(25))
    print(validate_age(15))
except ValueError as e:
    print("Error:", e)
```

### Q20
File handling aur Exception handling ko combine karein: Ek file read karein, agar file na mile toh error catch karein, aur `finally` block mein file closure handle karein.

```python
f = None
try:
    f = open("missing_file.txt", "r")
    print(f.read())
except FileNotFoundError:
    print("Error: File nahi mili!")
finally:
    if f is not None:          # sirf tab close karein jab file khuli ho
        f.close()
        print("File close ho gayi.")
    else:
        print("File khuli hi nahi thi, is liye close ki zaroorat nahi.")
```
